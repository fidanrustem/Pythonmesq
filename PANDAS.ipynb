```python
"""""
4. Select Rows Using iloc
Select the first 3 rows.
Select the last 2 rows.
Select the row at index 2.

5. Select Rows Using loc
Select the row where "Name" is "Charlie".
Select the rows where "Salary" > 60000.
Select "Name" and "City" columns for rows where "Age" > 30.

6. Boolean Masking
Select rows where "Age" < 30.
Select rows where "City" is "Miami".
Select rows where "Salary" > 55000 and "Age" < 40.
```

```python
import pandas as pd
data = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Eva"],
    "Age": [25, 30, 35, 40, 28],
    "City": ["New York", "Los Angeles", "Chicago", "Houston", "Miami"],
    "Salary": [50000, 60000, 70000, 80000, 55000]
}
df=pd.DataFrame(data)
print(df.iloc[:3])
print("----------------------------------------")
print(df.iloc[-2:])
print("----------------------------------------")
print(df.iloc[[2]])
print("----------------------------------------")
print(df.loc[df["Name"]=="Charlie"])
print(df.loc[df["Salary"]>60000])
print(df.loc[df["Age"]>30,["Name","City"]])
print("----------------------------------------")
print(df.loc[df["Age"]>30])
print(df.loc[df["City"]=="Miami"])
print(df.loc[(df["Salary"]==55000) & (df["Age"]<40)])
print("----------------------------------------")
```

```python
""""
7. Modify Data
Increase "Salary" by 5000 for all rows where "Age" > 30.
Change "City" of "Eva" to "Orlando".

8. Add New Column
Create a new column "Bonus" which is 10% of "Salary".
Create a column "Age Group": "Young" if Age < 30, "Adult" if Age ≥ 30.

9. Drop Columns or Rows
Drop the "Bonus" column.
Drop the row where "Name" is "Bob".

10. Sorting
Sort the DataFrame by "Age" ascending.
Sort the DataFrame by "Salary" descending.
""""
```

```python
import pandas as pd
data = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Eva"],
    "Age": [25, 30, 35, 40, 28],
    "City": ["New York", "Los Angeles", "Chicago", "Houston", "Miami"],
    "Salary": [50000, 60000, 70000, 80000, 55000]
}
df=pd.DataFrame(data)
df.loc[df["Age"]>30,"Salary"]=df.loc[df["Age"]>30,"Salary"]+5500
print(df)
df.loc[df["Name"]=="Eva",["City"]]="Orlando"
print(df)
print("----------------------------------------")
df["Bonus"]=df["Salary"]*0.10
print(df)
df["Age group"]="Adult"
df.loc[df["Age"]<30,"Age group"]="Young"
print(df)
print("----------------------------------------")
df=df.drop(columns=["Bonus"])
df=df.drop(df[df["Name"]=="Bob"].index)
print(df)
print("----------------------------------------")
print(df.sort_values(by="Age"))
print(df.sort_values(by="Salary",ascending=False))
```
