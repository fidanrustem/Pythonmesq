```python
""""
12. Given a 2D array of points, compute the Euclidean distance between each pair.
13. Apply a custom function f(x) = x² + 2x + 1 to every element using vectorization.
14. Given a matrix, return only its unique rows.
15. Implement 1D convolution manually between two arrays.
```

```python
parray=np.array([[0,2],[3,4],[6,8]])
for i in range(len(parray)):
  for j in range(i+1,len(parray)):
        d = np.linalg.norm(parray[i]-parray[j])
        print(d)

arr=np.array([1,2,3,4])
result=arr**2+2*arr+1
print(result)

matrix=np.array([[1,2],[3,4],[1,2],[5,6]])
unique=np.unique(matrix,axis=0)
print(unique)

a=np.array([1,2,3])
b=np.array([4,5])
n=len(a)
m=len(b)
result=np.zeros(n+m-1)
for i in range(n):
    for j in range(m):
        result[i+j]+=a[i]*b[j]
print(result)
```
