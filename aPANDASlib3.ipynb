```python
import pandas as pd
import numpy as np

np.random.seed(42)
data = pd.DataFrame({
    "EmployeeID": np.arange(1, 31),
    "Name": [f"Emp{i}" for i in range(1, 31)],
    "Department": np.random.choice(["HR", "Sales", "IT", "Finance"], size=30),
    "JoiningDate": pd.to_datetime(np.random.choice(pd.date_range("2018-01-01", "2023-01-01"), size=30)),
    "Salary": np.random.randint(40000, 120000, size=30),
    "Bonus": np.random.randint(2000, 15000, size=30),
    "ExperienceYears": np.random.randint(1, 15, size=30),
    "ProjectsCompleted": np.random.randint(0, 30, size=30),
    "RemoteDays": np.random.randint(0, 50, size=30)
})
data.groupby("Department").agg(
    AverageSalary=("Salary","mean"),
    TotalBonus=("Bonus","sum"),
    AvergaeExperience=("ExperienceYears","mean")
)
davg=data.groupby("Department")["ProjectsCompleted"].mean()
print(davg)
dmax=data.groupby("Department")["Salary"].max()
dmin=data.groupby("Department")["Salary"].min()
print(dmin,dmax)
dd=data.groupby(["ExperienceYears","Department"])["Salary"].mean()
print(dd)

data["DepartmentAvgSalary"]=data.groupby("Department")["Salary"].transform("mean")
print(data)
findwork=data[data["Salary"]>data["DepartmentAvgSalary"]]
print(findwork)
data=data.sort_values("JoiningDate")
data["DataCumSum"]=data.groupby("Department")["ProjectsCompleted"].cumsum()
print(data["DataCumSum"])

data["TotalCompensation"]=data["Salary"]+data["Bonus"]
data["Rank"]=data.groupby("Department")["TotalCompensation"].rank(ascending=False)
print(data[["Rank","TotalCompensation","Department","ProjectsCompleted","JoiningDate"]])
print("---------------------------------------------------------------------------------")
department_total=data.groupby("Department")["ProjectsCompleted"].transform("sum")
data["ProjectPercentage"]=data["ProjectsCompleted"]/department_total
print(data[["ProjectPercentage","Department"]])
data_25percent=data.groupby("Department")["Salary"].transform(
    lambda x:x.quantile(0.75)
)
data[data["Salary"]>=data_25percent]
```

```python
import pandas as pd
import numpy as np

np.random.seed(42)
data = pd.DataFrame({
    "EmployeeID": np.arange(1, 31),
    "Name": [f"Emp{i}" for i in range(1, 31)],
    "Department": np.random.choice(["HR", "Sales", "IT", "Finance"], size=30),
    "JoiningDate": pd.to_datetime(np.random.choice(pd.date_range("2018-01-01", "2023-01-01"), size=30)),
    "Salary": np.random.randint(40000, 120000, size=30),
    "Bonus": np.random.randint(2000, 15000, size=30),
    "ExperienceYears": np.random.randint(1, 15, size=30),
    "ProjectsCompleted": np.random.randint(0, 30, size=30),
    "RemoteDays": np.random.randint(0, 50, size=30)
})
data["ProductivityScore"]=data["ProjectsCompleted"]/data["ExperienceYears"]
print(data)
top5employees=data.sort_values(by="ProductivityScore",ascending=False)
top5employees=top5employees.head(5)
print(top5employees)
datanalyze=data.groupby("ExperienceYears")["Salary"].mean()
lowExperience=data["ExperienceYears"].quantile(0.25)
highSalary=data["Salary"].quantile(0.75)
data[(data["ExperienceYears"]<=lowExperience) & (data["Salary"]>=highSalary)]

data.sort_values(by="JoiningDate",ascending=True)
data["CumulativeData"]=data.groupby("Department")["ProjectsCompleted"].cumsum()
print(data.sort_values(by="CumulativeData"))
bins=[40000,60000,80000,100000,120000]
labels=[
    "40k-60k",
    "60k-80k",
    "80k-100k",
    "100k-120k"
]
data["SalaryRange"]=pd.cut(
    data["Salary"],
    bins=bins,
    labels=labels
)
print(data["SalaryRange"].value_counts())

data["ProductivityScore"]=data["ProjectsCompleted"]/data["ExperienceYears"]
departmentavgsal=data.groupby("Department")["Salary"].transform("mean")
data[(data["ProductivityScore"]>1.5) & (data["Salary"]>departmentavgsal)]
corr=data[
    ["Salary",
    "Bonus",
    "ExperienceYears",
    "ProjectsCompleted",
    "RemoteDays"]
].corr()
print(corr)

joined=data.groupby(
    data["JoiningDate"].dt.year
).size()
print(joined)
print(joined.idxmax())
limit=data["TotalCompensation"].quantile(0.90)
result=data[
    (data["TotalCompensation"]>limit) & (data["ProductivityScore"]>1.2)
]
print(result)
result = data[
    (data["ExperienceYears"] > 10) &
    (data["ProjectsCompleted"] < 5) &
    (data["RemoteDays"] > 30)
]
print(result)
```
