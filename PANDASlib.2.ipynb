```python
""""
1.Group by "Salesperson" and calculate the total sales per salesperson.
2. Average Sales by Region
Using the same DataFrame, group by "Region" and calculate the average sales.
3. Count of Sales per Salesperson
Group by "Salesperson" and count how many sales records each person has.
4. Maximum Sale per Region
Find the maximum sales value in each region.
```

```python
import pandas as pd
data={
    "Salesperson": ["Alice", "Bob", "Alice", "David", "Bob", "Alice"],
    "Region": ["East", "West", "East", "South", "West", "East"],
    "Sales": [200, 150, 300, 400, 250, 100]
}
df=pd.DataFrame(data)
datagroupBy=df.groupby("Salesperson")["Sales"].sum().to_frame()
print(datagroupBy)
print("----------------------------------------")
dataRegion=df.groupby("Region")["Sales"].mean().to_frame()
print(dataRegion)
print("----------------------------------------")
datacount=df.groupby("Salesperson")["Sales"].count().to_frame()
print(datacount)
print("----------------------------------------")
dataMaxim=df.groupby("Region")["Sales"].max().to_frame()
print(dataMaxim)
```

```python
"""""
5. Multiple Aggregations
For each "Salesperson", calculate total sales, average sales, and count of sales using a single groupby with agg().
6. Grouping with Filtering
Group by "Salesperson" and calculate total sales, then filter only those whose total sales are greater than 400.
7. Grouping by Multiple Columns
Group by "Region" and "Salesperson" and calculate total sales for each combination.
```

```python
import pandas as pd
data={
    "Salesperson": ["Alice", "Bob", "Alice", "David", "Bob", "Alice"],
    "Region": ["East", "West", "East", "South", "West", "East"],
    "Sales": [200, 150, 300, 400, 250, 100]
}
df=pd.DataFrame(data)
dataAgg=df.groupby("Salesperson").agg(
    TotalSales=("Sales","sum"),
    AverageSales=("Sales","mean"),
    CountSales=("Sales","count")
)
print(dataAgg)
print("----------------------------------------")
dataFiltered=df.groupby("Salesperson")["Sales"].sum().to_frame()
dataFiltered=dataFiltered[dataFiltered>400]
print(dataFiltered)
print("----------------------------------------")
dataMulti=df.groupby(["Region","Salesperson"])["Sales"].sum().to_frame()
print(dataMulti)
```

```python
"""""
8. Add Aggregated Values Back to Original DataFrame
Create a new column "Total_Sales_By_Salesperson" in the original DataFrame showing each salesperson’s total sales (use transform).
9. Percentage Contribution
For each salesperson, calculate their percentage contribution to the total sales.
10. Custom Aggregation
Group by "Region" and calculate a custom metric: difference between max and min sales (max - min) in that region.
```

```python
import pandas as pd
data={
    "Salesperson": ["Alice", "Bob", "Alice", "David", "Bob", "Alice"],
    "Region": ["East", "West", "East", "South", "West", "East"],
    "Sales": [200, 150, 300, 400, 250, 100]
}
df=pd.DataFrame(data)
df["Total_Sales_By_Salesperson"]=df.groupby("Salesperson")["Sales"].transform("sum")
print(df)
print("----------------------------------------")
TotalSales=df["Sales"].sum()
DataContribute=df.groupby("Salesperson")["Sales"].sum()/TotalSales*100
DataContribute=DataContribute.to_frame
print(DataContribute,"%")
print("----------------------------------------")
DaResult=df.groupby("Region")["Sales"].agg(lambda x:x.max()-x.min())
print(DaResult)
```
