```python
import pandas as pd
import numpy as np
import seaborn as sns
np.random.seed(42)
employee_data = pd.DataFrame({
    "EmployeeID": np.arange(1, 21),
    "Name": [f"Emp{i}" for i in range(1, 21)],
    "Department": np.random.choice(["HR", "Sales", "IT", "Finance"], size=20),
    "Salary": np.random.randint(40000, 120000, size=20),
    "ProjectsCompleted": np.random.randint(0, 20, size=20),
    "ExperienceYears": np.random.randint(1, 15, size=20),
    "RemoteDays": np.random.randint(0, 50, size=20)
})
sales_data = pd.DataFrame({
    "Month": ["Jan","Feb","Mar","Apr","May","Jun","Jul","Aug","Sep","Oct"],
    "Product_A": [250,300,400,350,500,450,480,470,510,550],
    "Product_B": [200,220,330,300,360,390,400,420,450,480],
    "Product_C": [150,180,210,200,230,250,260,280,300,320]
})
import matplotlib.pyplot as plt
sales=sales_data.melt( id_vars="Month",var_name="Product",value_name="Sales")
sns.lineplot(data=sales,x="Month",y="Sales",hue="Product",marker="o")
plt.show()
sns.barplot(data=sales,x="Month",y="Sales",hue="Product")
plt.show()
employee_data["RemoteCategory"] = pd.cut(employee_data["RemoteDays"],bins=[0,20,50],labels=["Low","High"])
sns.boxplot(data=employee_data,x="Department",y="Salary",hue="RemoteCategory")
plt.show()
sns.violinplot(data=employee_data,x="Department",y="Salary")
plt.show()
sns.scatterplot(data=employee_data,x="ExperienceYears",y="ProjectsCompleted",hue="Department",size="Salary")
plt.show()
g=sns.FacetGrid(employee_data,col="Department")
g.map(sns.histplot,"Salary")
```

```python
import pandas as pd
import numpy as np
import seaborn as sns
np.random.seed(42)
employee_data = pd.DataFrame({
    "EmployeeID": np.arange(1, 21),
    "Name": [f"Emp{i}" for i in range(1, 21)],
    "Department": np.random.choice(["HR", "Sales", "IT", "Finance"], size=20),
    "Salary": np.random.randint(40000, 120000, size=20),
    "ProjectsCompleted": np.random.randint(0, 20, size=20),
    "ExperienceYears": np.random.randint(1, 15, size=20),
    "RemoteDays": np.random.randint(0, 50, size=20)
})
sales_data = pd.DataFrame({
    "Month": ["Jan","Feb","Mar","Apr","May","Jun","Jul","Aug","Sep","Oct"],
    "Product_A": [250,300,400,350,500,450,480,470,510,550],
    "Product_B": [200,220,330,300,360,390,400,420,450,480],
    "Product_C": [150,180,210,200,230,250,260,280,300,320]
})
import matplotlib.pyplot as plt
corr = employee_data[
    [
        "Salary",
        "ProjectsCompleted",
        "ExperienceYears",
        "RemoteDays"
    ]
].corr()
sns.heatmap(corr,annot=True,cmap="coolwarm")
plt.show()
sns.regplot(data=employee_data,x="ExperienceYears",y="Salary")
plt.show()
sns.jointplot(data=employee_data,x="Salary",y="ProjectsCompleted")
sns.pairplot(
    employee_data,
    vars=[
        "Salary",
        "ExperienceYears",
        "ProjectsCompleted",
        "RemoteDays"
    ],hue="Department"
)
sales = sales_data.melt(id_vars="Month",var_name="Product",value_name="Sales")
pivot = sales.pivot(index="Month",columns="Product",values="Sales")
sns.heatmap(pivot,annot=True)
plt.show()
sns.swarmplot(data=employee_data,x="Department",y="Salary")
plt.show()
sns.kdeplot(data=employee_data,x="Salary",hue="Department",fill=True)
plt.show()
sales_data["A"]=sales_data["Product_A"].cumsum()
sales_data["B"]=sales_data["Product_B"].cumsum()
sales_data["C"]=sales_data["Product_C"].cumsum()
sns.lineplot(data=sales_data,x="Month",y="A",marker="o")
sns.lineplot(data=sales_data,x="Month",y="B",marker="o")
sns.lineplot(data=sales_data,x="Month",y="C",marker="o")
plt.show()
employee_data["Productivity"] = (employee_data["ProjectsCompleted"] /employee_data["ExperienceYears"])
top=employee_data.nlargest(3,"Productivity")
sns.scatterplot(data=employee_data,x="Salary",y="ProjectsCompleted")
plt.scatter(top["Salary"],top["ProjectsCompleted"],s=200)
plt.show()
avg=employee_data.groupby("Department").agg(Salary=("Salary","mean"),Projects=("ProjectsCompleted","mean")).reset_index()
fig,ax1=plt.subplots()
sns.barplot(data=avg,x="Department",y="Salary",ax=ax1)
ax2=ax1.twinx()
ax2.scatter(avg["Department"],avg["Projects"])
plt.show()
```
